### Exercise 15 — Vectorize a Function

**Question:** Convert the scalar function `maxx` below to work on two arrays.


```python
# Input — do not modify
def maxx(x, y):
    """Get the maximum of two items"""
    if x >= y:
        return x
    else:
        return y

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])
```

```python
# Write your code here:
result=np.vectorize(maxx)(a,b)
print(result)
```

### Exercise 42 — Probabilistic Sampling

**Question:** Randomly sample iris's species such that **setosa** appears **twice** as often as versicolor and virginica.

*# Make sure you have run Exercise 25 first.*


```python
# Write your code here:
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/iris.csv'
iris_data_clean = np.genfromtxt(url, delimiter=',', dtype='object')
species_all = iris_data_clean[:, 4]

# Assign weights: setosa=2, others=1
probs = np.where(species_all == b'Iris-setosa', 2, 1).astype(float)
probs /= probs.sum()

sampled = np.random.choice(species_all, size=100, replace=True, p=probs)
vals, counts = np.unique(sampled, return_counts=True)
for v, c in zip(vals, counts):
    print(v, ':', c)
```
